```python
# release_date sütununu tarih formatına çeviriyoruz
df_clean["release_date"] = pd.to_datetime(
    df_clean["release_date"],
    format="%m/%d/%y",
    errors="coerce"
)

# İki haneli yıl nedeniyle 2060, 2070 gibi yanlış yorumlanan
# eski tarihleri 100 yıl geriye çekiyoruz
future_date_mask = df_clean["release_date"].dt.year > 2026

df_clean.loc[future_date_mask, "release_date"] = (
    df_clean.loc[future_date_mask, "release_date"]
    - pd.DateOffset(years=100)
)

# Tarihten yeni özellikler çıkarıyoruz
df_clean["release_year"] = df_clean["release_date"].dt.year
df_clean["release_month"] = df_clean["release_date"].dt.month
df_clean["release_day"] = df_clean["release_date"].dt.day
df_clean["release_weekday"] = df_clean["release_date"].dt.dayofweek

# Hafta sonu bilgisi
# 0: Pazartesi, 1: Salı, ..., 5: Cumartesi, 6: Pazar
df_clean["is_weekend"] = (
    df_clean["release_weekday"].isin([5, 6])
).astype(int)

# Mevsim bilgisi
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    elif month in [9, 10, 11]:
        return "Autumn"
    else:
        return np.nan


df_clean["release_season"] = (
    df_clean["release_month"]
    .apply(get_season)
)

# Kontrol
print("En eski yıl:", df_clean["release_year"].min())
print("En yeni yıl:", df_clean["release_year"].max())
print("Eksik tarih sayısı:", df_clean["release_date"].isnull().sum())

df_clean[
    [
        "release_date",
        "release_year",
        "release_month",
        "release_day",
        "release_weekday",
        "is_weekend",
        "release_season"
    ]
].head(10)
```

```python
import ast

# JSON benzeri metinleri güvenli şekilde listeye çeviren fonksiyon
def parse_list(value):
    if pd.isna(value):
        return []
    
    try:
        parsed_value = ast.literal_eval(value)
        
        if isinstance(parsed_value, list):
            return parsed_value
        
        return []
    
    except (ValueError, SyntaxError, TypeError):
        return []


# İlgili sütunları liste yapısına çeviriyoruz
list_columns = [
    "genres",
    "production_companies",
    "production_countries",
    "spoken_languages",
    "cast",
    "crew"
]

for column in list_columns:
    df_clean[column + "_parsed"] = df_clean[column].apply(parse_list)


# Liste uzunluklarından sayısal özellikler çıkarıyoruz
df_clean["genre_count"] = df_clean["genres_parsed"].apply(len)

df_clean["production_company_count"] = (
    df_clean["production_companies_parsed"].apply(len)
)

df_clean["production_country_count"] = (
    df_clean["production_countries_parsed"].apply(len)
)

df_clean["spoken_language_count"] = (
    df_clean["spoken_languages_parsed"].apply(len)
)

df_clean["cast_count"] = df_clean["cast_parsed"].apply(len)

df_clean["crew_count"] = df_clean["crew_parsed"].apply(len)


# Bir film serisine veya koleksiyonuna ait mi?
df_clean["belongs_to_collection_flag"] = (
    df_clean["belongs_to_collection"].notna()
).astype(int)


# İlk türü ana tür olarak çıkarıyoruz
def get_main_genre(parsed_genres):
    if len(parsed_genres) > 0:
        return parsed_genres[0].get("name", "Unknown")
    
    return "Unknown"


df_clean["main_genre"] = (
    df_clean["genres_parsed"]
    .apply(get_main_genre)
)


# Kontrol
df_clean[
    [
        "genres",
        "main_genre",
        "genre_count",
        "production_company_count",
        "production_country_count",
        "spoken_language_count",
        "cast_count",
        "crew_count",
        "belongs_to_collection_flag"
    ]
].head(10)
```

```python
# Modelde kullanılacak sayısal özellikler
numeric_features = [
    "log_budget",
    "budget_missing",
    "runtime",
    "release_year",
    "release_month",
    "release_weekday",
    "is_weekend",
    "genre_count",
    "production_company_count",
    "production_country_count",
    "spoken_language_count",
    "cast_count",
    "crew_count",
    "belongs_to_collection_flag"
]

# Modelde kullanılacak kategorik özellikler
categorical_features = [
    "original_language_grouped",
    "main_genre_grouped",
    "release_season"
]

# Filmleri çıkış tarihine göre sıralıyoruz
df_model = df_clean.sort_values("release_date").reset_index(drop=True)

# Model girdileri
X = df_model[numeric_features + categorical_features].copy()

# Hedef değişken: logaritmik gişe geliri
y = np.log1p(df_model["revenue"])

# Verinin %80'i eğitim, %20'si test
split_index = int(len(df_model) * 0.80)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

# Gerçek dolar cinsinden gelirleri de saklıyoruz
revenue_train = df_model["revenue"].iloc[:split_index].copy()
revenue_test = df_model["revenue"].iloc[split_index:].copy()

print("Eğitim veri boyutu:", X_train.shape)
print("Test veri boyutu:", X_test.shape)

print("\nEğitim dönemi:")
print(df_model["release_date"].iloc[:split_index].min())
print(df_model["release_date"].iloc[:split_index].max())

print("\nTest dönemi:")
print(df_model["release_date"].iloc[split_index:].min())
print(df_model["release_date"].iloc[split_index:].max())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Sayısal ve kategorik sütunlara farklı işlemler uyguluyoruz
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

preprocessor
```

```python
from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    Lasso,
    ElasticNet
)

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    AdaBoostRegressor,
    GradientBoostingRegressor,
    HistGradientBoostingRegressor
)

models = {
    "Linear Regression": LinearRegression(),

    "Ridge": Ridge(
        alpha=1.0
    ),

    "Lasso": Lasso(
        alpha=0.001,
        max_iter=20000
    ),

    "ElasticNet": ElasticNet(
        alpha=0.001,
        l1_ratio=0.5,
        max_iter=20000
    ),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "AdaBoost": AdaBoostRegressor(
        n_estimators=100,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        random_state=42
    ),

    "HistGradientBoosting": HistGradientBoostingRegressor(
        max_iter=100,
        random_state=42
    )
}

print("Karşılaştırılacak model sayısı:", len(models))
```

```python
import time
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

results = []
trained_pipelines = {}

for model_name, model in models.items():

    print(f"{model_name} eğitiliyor...")

    # Her model için ön işleme + modelden oluşan Pipeline
    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    # Eğitim süresini ölçüyoruz
    training_start = time.perf_counter()

    pipeline.fit(X_train, y_train)

    training_time = time.perf_counter() - training_start

    # Tahmin süresini ölçüyoruz
    prediction_start = time.perf_counter()

    train_predictions_log = pipeline.predict(X_train)
    test_predictions_log = pipeline.predict(X_test)

    prediction_time = time.perf_counter() - prediction_start

    # Log ölçekte sonuçlar
    train_r2 = r2_score(
        y_train,
        train_predictions_log
    )

    test_r2 = r2_score(
        y_test,
        test_predictions_log
    )

    log_mae = mean_absolute_error(
        y_test,
        test_predictions_log
    )

    log_rmse = np.sqrt(
        mean_squared_error(
            y_test,
            test_predictions_log
        )
    )

    # Tahminleri tekrar dolar cinsine çeviriyoruz
    train_predictions_revenue = np.expm1(
        train_predictions_log
    )

    test_predictions_revenue = np.expm1(
        test_predictions_log
    )

    # Negatif gelir tahminlerini 0'a çekiyoruz
    train_predictions_revenue = np.maximum(
        train_predictions_revenue,
        0
    )

    test_predictions_revenue = np.maximum(
        test_predictions_revenue,
        0
    )

    # Gerçek dolar ölçeğindeki hata değerleri
    revenue_mae = mean_absolute_error(
        revenue_test,
        test_predictions_revenue
    )

    revenue_rmse = np.sqrt(
        mean_squared_error(
            revenue_test,
            test_predictions_revenue
        )
    )

    # Train ve test R² farkı
    overfitting_gap = train_r2 - test_r2

    results.append({
        "Model": model_name,
        "Train R²": train_r2,
        "Test R²": test_r2,
        "R² Farkı": overfitting_gap,
        "Log MAE": log_mae,
        "Log RMSE": log_rmse,
        "Gelir MAE ($)": revenue_mae,
        "Gelir RMSE ($)": revenue_rmse,
        "Eğitim Süresi (sn)": training_time,
        "Tahmin Süresi (sn)": prediction_time
    })

    # Daha sonra kullanabilmek için eğitilen modeli saklıyoruz
    trained_pipelines[model_name] = pipeline


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Test R²",
    ascending=False
).reset_index(drop=True)

results_df
```

```python
formatted_results = results_df.copy()

percentage_columns = [
    "Train R²",
    "Test R²",
    "R² Farkı",
    "Log MAE",
    "Log RMSE"
]

formatted_results[percentage_columns] = (
    formatted_results[percentage_columns]
    .round(4)
)

money_columns = [
    "Gelir MAE ($)",
    "Gelir RMSE ($)"
]

formatted_results[money_columns] = (
    formatted_results[money_columns]
    .round(0)
    .astype("int64")
)

time_columns = [
    "Eğitim Süresi (sn)",
    "Tahmin Süresi (sn)"
]

formatted_results[time_columns] = (
    formatted_results[time_columns]
    .round(4)
)

formatted_results
```

```python
def get_feature_group(feature_name):
    if feature_name.startswith("original_language_grouped_"):
        return "original_language"

    if feature_name.startswith("main_genre_grouped_"):
        return "main_genre"

    if feature_name.startswith("release_season_"):
        return "release_season"

    return feature_name


feature_importance_df["Özellik Grubu"] = (
    feature_importance_df["Özellik"]
    .apply(get_feature_group)
)

grouped_importance_df = (
    feature_importance_df
    .groupby(
        "Özellik Grubu",
        as_index=False
    )["Önem"]
    .sum()
    .sort_values(
        by="Önem",
        ascending=False
    )
    .reset_index(drop=True)
)

grouped_importance_df
```

```python
# Barbie'nin gerçek dünya çapı gişe geliri
barbie_actual_revenue = 1_447_685_476

barbie_model_results = []

# trained_pipelines içindeki bütün eğitilmiş modelleri sırayla deniyoruz
for model_name, pipeline in trained_pipelines.items():

    # Model logaritmik gişe tahmini üretir
    predicted_log_revenue = pipeline.predict(
        barbie_movie
    )[0]

    # Logaritmik tahmini dolar cinsine çeviriyoruz
    predicted_revenue = np.expm1(
        predicted_log_revenue
    )

    # Gelir negatif olamayacağı için güvenlik kontrolü
    predicted_revenue = max(
        predicted_revenue,
        0
    )

    # Gerçek değerle tahmin arasındaki mutlak fark
    absolute_error = abs(
        barbie_actual_revenue
        - predicted_revenue
    )

    # Gerçek değere göre yüzde hata
    percentage_error = (
        absolute_error
        / barbie_actual_revenue
    ) * 100

    # Model yüksek mi düşük mü tahmin etti?
    if predicted_revenue > barbie_actual_revenue:
        prediction_direction = "Yüksek tahmin"
    elif predicted_revenue < barbie_actual_revenue:
        prediction_direction = "Düşük tahmin"
    else:
        prediction_direction = "Tam tahmin"

    barbie_model_results.append({
        "Model": model_name,
        "Tahmini Gişe ($)": predicted_revenue,
        "Gerçek Gişe ($)": barbie_actual_revenue,
        "Mutlak Hata ($)": absolute_error,
        "Yüzde Hata (%)": percentage_error,
        "Tahmin Yönü": prediction_direction
    })


# Sonuçları tabloya çeviriyoruz
barbie_models_df = pd.DataFrame(
    barbie_model_results
)

# En düşük yüzde hataya göre sıralıyoruz
barbie_models_df = (
    barbie_models_df
    .sort_values(
        "Yüzde Hata (%)",
        ascending=True
    )
    .reset_index(drop=True)
)

# Milyon dolar cinsinden okunabilir sütunlar
barbie_models_df[
    "Tahmini Gişe (Milyon $)"
] = (
    barbie_models_df[
        "Tahmini Gişe ($)"
    ] / 1_000_000
).round(2)

barbie_models_df[
    "Gerçek Gişe (Milyon $)"
] = (
    barbie_models_df[
        "Gerçek Gişe ($)"
    ] / 1_000_000
).round(2)

barbie_models_df[
    "Mutlak Hata (Milyon $)"
] = (
    barbie_models_df[
        "Mutlak Hata ($)"
    ] / 1_000_000
).round(2)

barbie_models_df[
    "Yüzde Hata (%)"
] = (
    barbie_models_df[
        "Yüzde Hata (%)"
    ].round(2)
)

barbie_models_df[
    [
        "Model",
        "Tahmini Gişe (Milyon $)",
        "Gerçek Gişe (Milyon $)",
        "Mutlak Hata (Milyon $)",
        "Yüzde Hata (%)",
        "Tahmin Yönü"
    ]
]
```
